```python
import random

def kmeans(data, k=2, max_iters=10):
    centroids = random.sample(data, k)
    for _ in range(max_iters):
        clusters = [[] for _ in range(k)]
        for x in data:
            distances = [abs(x - c) for c in centroids]
            cluster_idx = distances.index(min(distances))
            clusters[cluster_idx].append(x)
        
        centroids = [sum(c) / len(c) if c else random.choice(data) for c in clusters]
    return centroids, clusters

weights = [3.2, 4.1, 3.8, 4.5, 15.2, 22.0, 18.5, 25.1, 3.6, 20.0]
centroids, clusters = kmeans(weights)

sorted_centroids = sorted(centroids)
cat_centroid = sorted_centroids[0]
dog_centroid = sorted_centroids[1]

test_weight = 16.5
assigned_cluster = min(centroids, key=lambda c: abs(c - test_weight))

print(f"Cat cluster center: {cat_centroid}")
print(f"Dog cluster center: {dog_centroid}")
print(f"Weight {test_weight} is classified as: {'Cat' if assigned_cluster == cat_centroid else 'Dog'}")
```
