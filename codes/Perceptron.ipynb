```python
class Perceptron:
    def __init__(self, input_size):
        self.weights = [0.0 for _ in range(input_size)]
        self.bias = 0.0
        self.lr = 0.1

    def predict(self, inputs):
        total = sum(w * i for w, i in zip(self.weights, inputs)) + self.bias
        return 1 if total >= 0 else 0

    def train(self, training_inputs, labels, epochs):
        for _ in range(epochs):
            for inputs, label in zip(training_inputs, labels):
                prediction = self.predict(inputs)
                error = label - prediction
                self.weights = [w + self.lr * error * i for w, i in zip(self.weights, inputs)]
                self.bias += self.lr * error

training_data = [
    [5000, 700],
    [2000, 500],
    [8000, 750],
    [1500, 400],
    [6000, 650]
]
labels = [1, 0, 1, 0, 1]

p = Perceptron(2)
p.train(training_data, labels, 20)

person = [5500, 680]
result = p.predict(person)
print("deserve a lone" if result == 1 else "not deserve a lone")
```
